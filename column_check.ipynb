##### Check: AGMT_END_YMD (Date, with 2 sentinel values: '정보없음', '무약정')

```python
agmt_end_ymd_summary = duckdb.sql(f"""
    SELECT
        CASE
            WHEN AGMT_END_YMD = '정보없음' THEN '정보없음'
            WHEN AGMT_END_YMD = '무약정' THEN '무약정'
            ELSE '실제 날짜'
        END AS value_type,
        count(*) AS n,
        min(TRY_STRPTIME(AGMT_END_YMD, '%Y%m%d')) AS min_date,
        max(TRY_STRPTIME(AGMT_END_YMD, '%Y%m%d')) AS max_date
    FROM read_csv([{file_list_sql}], all_varchar=true, union_by_name=true)
    GROUP BY value_type
    ORDER BY n DESC
""").df()

agmt_end_ymd_summary
```
